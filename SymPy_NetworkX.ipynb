## Operaciones con matrices utilizando SymPy

La librería SymPy permite trabajar con matrices simbólicas y realizar operaciones de álgebra lineal de forma exacta.

Consideremos la siguiente matriz

$$
A=
\begin{pmatrix}
2 & -1 & 0 & 3\\
1 & 4 & 2 & -2\\
0 & 1 & 3 & 1\\
2 & 0 & -1 & 5
\end{pmatrix}
$$

y el vector

$$
v=
\begin{pmatrix}
1\\
2\\
-1\\
3
\end{pmatrix}
$$

El producto está definido como

$$
Av
$$

y produce un nuevo vector en $\mathbb{R}^4$.

```python
import sympy as sp

A = sp.Matrix([
[2,-1,0,3],
[1,4,2,-2],
[0,1,3,1],
[2,0,-1,5]
])

v = sp.Matrix([
[1],
[2],
[-1],
[3]
])

A*v
```

## Potencias de matrices

Para una matriz cuadrada $A$, la potencia $A^n$ representa la multiplicación repetida de la matriz por sí misma

$$
A^n = A \cdot A \cdots A
$$



```python
A**2
```

```python
A**5
```

## Solución de sistemas lineales

$$

El objetivo es encontrar el vector $x$ que satisface esta ecuación.

## Valores propios y vectores propios

Los valores propios de una matriz $A$ se obtienen resolviendo

$$
\det(A-\lambda I)=0
$$

Los vectores propios asociados satisfacen

$$
Av=\lambda v
$$

```python
A.eigenvals()
```

## Matriz de transición

Consideremos una cadena de Markov con cinco estados

$$
S_1,S_2,S_3,S_4,S_5
$$

y la matriz de transición

$$
P=
\begin{pmatrix}
0.3 & 0.4 & 0.3 & 0 & 0\\
0.2 & 0.5 & 0.3 & 0 & 0\\
0 & 0 & 1 & 0 & 0\\
0.4 & 0 & 0 & 0.6 & 0\\
0 & 0 & 0 & 0.2 & 0.8
\end{pmatrix}
$$

Cada entrada $p_{ij}$ representa la probabilidad de transición del estado $S_i$ al estado $S_j$.

```python
import numpy as np

P = np.array([
[0.3,0.4,0.3,0,0],
[0.2,0.5,0.3,0,0],
[0,0,1,0,0],
[0.4,0,0,0.6,0],
[0,0,0,0.2,0.8]
])

P
```

La potencia $P^n$ describe la probabilidad de transición en $n$ pasos del sistema.

```python
np.linalg.matrix_power(P,5)
```

```python
import networkx as nx
import matplotlib.pyplot as plt

states = ["S1","S2","S3","S4","S5"]

G = nx.DiGraph()

for s in states:
    G.add_node(s)

for i in range(len(states)):
    for j in range(len(states)):
        if P[i,j] > 0:
            G.add_edge(states[i],states[j],weight=P[i,j])

pos = nx.spring_layout(G)

nx.draw(G,pos,with_labels=True,node_size=2500,node_color="lightblue")

labels = nx.get_edge_attributes(G,'weight')
nx.draw_networkx_edge_labels(G,pos,edge_labels=labels)

plt.show()
```
