# (a) Distribución Weibull We(α,1)

## Función de densidad

$$
f(x)=\alpha x^{\alpha-1}e^{-x^\alpha}
$$

## Función acumulada

$$
F(x)=\int_0^x \alpha t^{\alpha-1}e^{-t^\alpha}dt
$$

Sea:

$$
u=t^\alpha
$$
Entonces:

$$
du=\alpha t^{\alpha-1}dt
$$

Sustituyendo:

$$
F(x)=\int_0^{x^\alpha} e^{-u}du
$$

$$
F(x)=1-e^{-x^\alpha}
$$

## Método de la transformada inversa

$$
U=1-e^{-x^\alpha}
$$

$$
e^{-x^\alpha}=1-U
$$

$$
x^\alpha=-\ln(1-U)
$$

$$
X=(-\ln(1-U))^{1/\alpha}
$$

```python



alpha = 2


x = np.linspace(0,5,1000)

# Función de densidad Weibull
f = alpha*(x**(alpha-1))*np.exp(-(x**alpha))


plt.plot(x,f)

plt.title("Densidad Weibull")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid()

plt.show()
```

```python
np.mean(f)
```

```python
np.var(f)
```

```python
np.std(f)
```

```python


def weibull(alpha, n):

    muestras = []

    for i in range(n):

        U = rnd.random()

        x = (-np.log(1-U))**(1/alpha)

        muestras.append(x)

    return muestras
```

```python
W = weibull(2,1000)
```

```python
plt.hist(W,bins=50)
```

```python
np.mean(W)
```

```python
np.var(W)
```

# (b) Distribución Uniforme Gu(0,1)

## Función de densidad

$$
f(x)=1
$$

## Función acumulada

$$
F(x)=\int_0^x 1dt
$$

$$
F(x)=x
$$

## Método de la transformada inversa

$$

$$
X=U

# (c) Distribución Cauchy C(0,1)

## Función de densidad

$$
f(x)=\frac{1}{\pi(1+x^2)}
$$

## Función acumulada

$$
F(x)=\int_{-\infty}^{x}\frac{1}{\pi(1+t^2)}dt
$$

Sacando la constante:

$$
F(x)=\frac{1}{\pi}\int_{-\infty}^{x}\frac{1}{1+t^2}dt
$$

Sabemos que:

$$
\int \frac{1}{1+t^2}dt=\arctan(t)
$$

Entonces:

$$
F(x)=\frac{1}{\pi}[\arctan(t)]_{-\infty}^{x}
$$

Como:

$$
\arctan(-\infty)=-\frac{\pi}{2}
$$

obtenemos:

$$
F(x)=\frac{1}{\pi}\left(\arctan(x)+\frac{\pi}{2}\right)
$$

$$
F(x)=\frac{1}{\pi}\arctan(x)+\frac{1}{2}
$$

## Método de la transformada inversa

$$
U=\frac{1}{\pi}\arctan(x)+\frac{1}{2}
$$

$$
U-\frac{1}{2}=\frac{1}{\pi}\arctan(x)
$$

$$
\pi\left(U-\frac{1}{2}\right)=\arctan(x)
$$

Aplicando tangente:

$$
X=\tan\left(\pi\left(U-\frac{1}{2}\right)\right)
$$

```python

x = np.linspace(-10,10,1000)

# Función de densidad Cauchy
f = 1/(np.pi*(1+x**2))


plt.plot(x,f)

plt.title("Densidad Cauchy")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid()

plt.show()
```

```python
np.mean(f)
```

```python
np.var(f)
```

```python
np.std(f)
```

```python
def Cauchy(n):
  muestras=[]
  for i in range(n):
    U=rnd.random()
    x=np.tan(np.pi*(U-0.5))
    muestras.append(x)
  return muestras
```

```python
C = Cauchy(10000)
```

```python
plt.hist(C, bins=100, density=True, range=(-10,10))
```

```python
np.mean(C)
```

```python
np.var(C)
```

```python
np.std(C)
```

# (d) Distribución Logística L(0,1)

## Función de densidad

$$
f(x)=\frac{e^{-x}}{(1+e^{-x})^2}
$$

## Función acumulada

$$
F(x)=\int_{-\infty}^{x}\frac{e^{-t}}{(1+e^{-t})^2}dt
$$

Sea:

$$
u=1+e^{-t}
$$

Entonces:

$$
du=-e^{-t}dt
$$

Sustituyendo:

$$
F(x)=-\int \frac{1}{u^2}du
$$

$$
F(x)=\frac{1}{u}
$$

Regresando a la variable original:

$$
F(x)=\frac{1}{1+e^{-x}}
$$

## Método de la transformada inversa

$$
U=\frac{1}{1+e^{-x}}
$$

Invertimos:

$$
\frac{1}{U}=1+e^{-x}
$$

$$
\frac{1}{U}-1=e^{-x}
$$

$$
\frac{1-U}{U}=e^{-x}
$$

Aplicando logaritmo:

$$
-x=\ln\left(\frac{1-U}{U}\right)
$$

Finalmente:

$$
X=\ln\left(\frac{U}{1-U}\right)
$$

```python

x = np.linspace(-10,10,1000)

# Función de densidad logística
f = (np.exp(-x))/((1+np.exp(-x))**2)


plt.plot(x,f)

plt.title("Densidad Logística")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid()

plt.show()
```

```python
np.mean(f)
```

```python
np.var(f)
```

```python
np.std(f)
```

```python
def Logistica(n):
  muestras = []
  for i in range(n):
    U=rnd.random()
    x= np.log(U/(1-U))
    muestras.append(x)
  return muestras
```

```python
L = Logistica(1000)
```

```python
plt.hist(L, bins=50)
```

```python
np.mean(L)
```

```python
np.var(L)
```

```python
np.std(L)
```

# (e) Distribución Pareto Par(α)

## Función de densidad

$$
f(x)=\frac{\alpha}{x^{\alpha+1}}
$$

con:

$$
x\geq 1
$$

## Función acumulada

$$
F(x)=\int_1^x \frac{\alpha}{t^{\alpha+1}}dt
$$

Reescribiendo:

$$
F(x)=\alpha\int_1^x t^{-(\alpha+1)}dt
$$

Integrando:

$$
F(x)=\alpha\left[\frac{t^{-\alpha}}{-\alpha}\right]_1^x
$$

$$
F(x)=-[t^{-\alpha}]_1^x
$$

$$
F(x)=-(x^{-\alpha}-1)
$$

Finalmente:

$$
F(x)=1-\frac{1}{x^\alpha}
$$

## Método de la transformada inversa

$$
U=1-\frac{1}{x^\alpha}
$$

Despejando:

$$
\frac{1}{x^\alpha}=1-U
$$

$$
x^\alpha=\frac{1}{1-U}
$$

Aplicando potencia:

$$
X=\left(\frac{1}{1-U}\right)^{1/\alpha}
$$

```python

alpha = 3


x = np.linspace(1,10,1000)

# Función de densidad Pareto
f = alpha/(x**(alpha+1))


plt.plot(x,f)

plt.title("Densidad Pareto")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid()

plt.show()
```

```python
np.mean(f)
```

```python
np.var(f)
```

```python
np.std(f)
```

```python
def Par(alpha,n):
  muestras = []
  for i in range(n):
    U = rnd.random()
    x=(1/1-U)**(1/alpha)
    muestras.append(x)
  return muestras
```

```python
P = Par(3,10000)
```

```python
plt.hist(P,bins=50)
```

```python
np.mean(P)
```

```python
np.var(P)
```
