## Simulación de una cadena de Markov: tablero de colores

Se implementa un programa en Python para simular una **cadena de Markov** con cuatro estados que representan los colores de un tablero:

- 0 = Rojo  
- 1 = Azul  
- 2 = Verde  
- 3 = Amarillo  

Las probabilidades de transición entre los colores están dadas por la matriz

$$
P =
\begin{pmatrix}
0.25 & 0 & 0.25 & 0.5 \\
0.5 & 0.25 & 0 & 0.25 \\
0.25 & 0.5 & 0.25 & 0 \\
0 & 0.25 & 0.5 & 0.25
\end{pmatrix}.
$$

Cada entrada $P_{ij}$ representa la probabilidad de pasar del estado $i$ al estado $j$.

El algoritmo comienza definiendo una **distribución inicial uniforme**

$$
\alpha = (0.25,0.25,0.25,0.25),
$$

lo cual significa que inicialmente cada color tiene la misma probabilidad de ocurrir.

Posteriormente se genera un número aleatorio $r \in [0,1]$ para seleccionar el estado inicial de acuerdo con dicha distribución. Después, el programa simula la evolución de la cadena durante muchos pasos. En cada iteración se genera un número aleatorio y se utiliza la fila correspondiente de la matriz de transición para determinar el siguiente estado.

Para verificar el comportamiento a largo plazo de la cadena se utiliza un número grande de pasos (por ejemplo $10\,000$). Al finalizar la simulación se cuentan las veces que aparece cada estado y se calculan las probabilidades empíricas.

Debido a que la matriz de transición es **doblemente estocástica**, la distribución estacionaria es

$$
\pi = (1/4,1/4,1/4,1/4).
$$

Por lo tanto, a largo plazo se espera que los cuatro colores del tablero aparezcan con la **misma probabilidad**. La simulación confirma este resultado, ya que las frecuencias observadas de cada estado se aproximan a $0.25$ cuando el número de pasos es suficientemente grande.

```python
import random

# Matriz de transición del tablero de colores
# Estados:
# 0 = Rojo
# 1 = Azul
# 2 = Verde
# 3 = Amarillo


matriz = [
[0.25,0,0.25,0.5],
[0.5,0.25,0,0.25],
[0.25,0.5,0.25,0],
[0,0.25,0.5,0.25]
]



alfa = [0.25,0.25,0.25,0.25]

print("Distribución inicial:", alfa)




r = random.random()
suma = 0

for i in range(4):
    suma += alfa[i]
    if r <= suma:
        estado = i
        break




pasos = 10000

cadena = [estado]



for _ in range(pasos):

    r = random.random()
    suma = 0

    for j in range(4):
        suma += matriz[estado][j]
        if r <= suma:
            estado = j
            break

    cadena.append(estado)


print("Cadena:", cadena)




conteo = [0,0,0,0]

for estado in cadena:
    conteo[estado] += 1

print("\nFrecuencia de cada color:")
for i in range(4):
    print("Estado",i,":",conteo[i])




print("\nProbabilidad aproximada:")

total = len(cadena)

for i in range(4):
    print("Estado",i,":",conteo[i]/total)
```
